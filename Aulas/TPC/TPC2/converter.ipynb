**Método para ler e converter o texto .md**

```python
import re

def markconverter(text):
    #imagens
    text = re.sub(r'!\[(.*?)\]\((.*?)\)',r'^<img src="\2" alt="\1"/>',text)
    #links
    text = re.sub(r'\[(.*?)\]\((.*?)\)', r'^<a href="\2">\1</a>', text)
    #bold
    text = re.sub(r'\*\*(.*?)\*\*', r'<b>\1</b>', text)
    #destacado
    text = re.sub(r'\`(.*?)\`', r'<code>\1</code>', text)
    #itálico
    text = re.sub(r'\*(.*?)\*', r'<i>\1</i>', text)
    #header com '#'
    text = re.sub(r'#\s+(.+)$', r'<h1>\1</h1>', text, flags=re.M)
    #header com '##'
    text = re.sub(r'^##\s+(.+)$',r'<h2>\1</h2>' , text, flags=re.M)
    #header com '###'
    text = re.sub(r'^###\s+(.+)$',r'<h3>\1</h3>' , text, flags=re.M)
    #lista numerada
    numlist = [] #criamos a lista
    tokens = text.split('\n') #separamos a string .md em tokens
    inside = 0 #flag para noção de início e fim da lista
    for tkn in tokens:
        match = re.match(r'^\d+\.\s+(.*)$', tkn)
        if match:
            if not inside: # início da lista (na primeira vez que itera)
                numlist.append('<ol>')
                inside = 1 # qualquer outro token que advenha, nao faz append de '<ol>'
            numlist.append(f'<li>{match.group(1)}</li>')
        else:
            if inside:
                numlist.append('<ol>')
                inside = 0
            numlist.append(tkn) #se não for elemento da lista, é convetido na mesma para html, só que fora da ista

    if inside:
        numlist.append('<ol>') #lista vazia
    return '\n'.join(numlist)
```

```python
example = """
# Exemplo
Este é um **exemplo** ...
Este é um *exemplo* ...
1. Primeiro item
2. Segundo item
3. Terceiro item
Como pode ser consultado em [página da UC](http://www.uc.pt)
![imagem dum coelho](http://www.coellho.com) ...
`destacado`
"""

print(markconverter(example))
```
